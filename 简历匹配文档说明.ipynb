# 简历匹配

运行match.py，将简历json的resume_content取出来经过同样的embedding之后，用数据库提供的query方法找到相似度最高的10条数据，并获取它的元数据，即索引数据。

其实还可以根据元数据做筛选。

相似度计算默认为计算余弦相似度，值在-1到1，表述了1024维向量空间中，两个向量的夹角。越接近0，越相似。
输出match_results.json，放入我们想要的字段。

resume_chromadb_operater.py和job_chromadb_operater.py是增删查的工具

SRC文件夹下是爬虫工具。utils下是deepseek api调用工具类。

# OCR


OCR文件夹下是简历转图片工具，使用百度PaddleOCR，文本方向分类模型ch_ppocr_mobile_v2.0_cls_infer，中文检测模型ch_PP-OCRv4_det_infer，文本识别模型ch_PP-OCRv4_rec_infer，存在~文件夹下。
识别逻辑：二值化，版面分析，提取文字。
